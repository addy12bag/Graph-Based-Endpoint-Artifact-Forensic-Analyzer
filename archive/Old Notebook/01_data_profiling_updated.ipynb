```python
# Start with quarantine records explicitly labelled as malicious
malicious_quarantine = quarantine[
    quarantine["message"].str.contains(
        "FileType: Malicious",
        case=False,
        na=False
    )
].copy()

print("Raw malicious quarantine records:",
      len(malicious_quarantine))

print("Endpoints:",
      malicious_quarantine["endpoint_id"].nunique())
```

```python
# Remove known false-positive artifacts
malicious_clean = malicious_quarantine[
    ~malicious_quarantine["message"].str.contains(
        r"File:\s*recently-used\.xbel\b",
        case=False,
        na=False,
        regex=True
    )
].copy()

malicious_clean = malicious_clean[
    ~malicious_clean["message"].str.contains(
        r"File:\s*[^|]*\.desktop\b",
        case=False,
        na=False,
        regex=True
    )
].copy()

print("After false-positive filtering:",
      len(malicious_clean))

print("Endpoints:",
      malicious_clean["endpoint_id"].nunique())
```

```python
display(
    malicious_clean[
        ["endpoint_id", "event_time", "message"]
    ]
    .sort_values(["endpoint_id", "event_time"])
    .head(50)
)
```

```python
malicious_clean["file_name"] = (
    malicious_clean["message"]
    .str.extract(r"File:\s*(.*?)\s*\|", expand=False)
)

display(
    malicious_clean[
        ["endpoint_id", "file_name"]
    ].head(30)
)
```

```python
malicious_clean["sha256"] = (
    malicious_clean["message"]
    .str.extract(
        r"SHA256:\s*([A-Fa-f0-9]{64})",
        expand=False
    )
)

display(
    malicious_clean[
        ["endpoint_id", "file_name", "sha256"]
    ].head(30)
)
```

```python
unique_artifacts = (
    malicious_clean[
        ["endpoint_id", "file_name", "sha256"]
    ]
    .drop_duplicates()
    .copy()
)

print("Unique endpoint/file/hash combinations:",
      len(unique_artifacts))
```

```python
display(
    unique_artifacts.sort_values(
        ["endpoint_id", "file_name"]
    ).head(50)
)
```


## Malicious Security Telemetry

The endpoint-security dataset contains explicit security detections marked
`FileType: Malicious`. This section profiles those records before they are
represented in the endpoint–artifact graph.

This is detection evidence from the source telemetry. It should not by itself
be interpreted as proof that an endpoint was compromised or that a malicious
artifact was executed.


```python

# Load the endpoint-security source data independently so this profiling
# section remains reproducible even if earlier notebook cells are reordered.
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent
security_path = project_root / "data" / "endpoint_security.csv"

security_df = pd.read_csv(security_path)

# Restrict the analysis to records explicitly classified as malicious by the
# source security telemetry rather than inferring maliciousness from filenames.
malicious_security_df = security_df[
    security_df["message"]
    .fillna("")
    .str.contains(
        "FileType: Malicious",
        case=False,
        na=False
    )
].copy()

print(f"Security records: {len(security_df):,}")
print(f"Explicit malicious records: {len(malicious_security_df):,}")
print(
    "Endpoints with explicit malicious telemetry:",
    malicious_security_df["endpoint_id"].nunique()
)

display(
    malicious_security_df[
        [
            "endpoint_id",
            "event_time",
            "message"
        ]
    ].head(10)
)
```

```python
# Load the complete raw endpoint-security dataset independently of the
# malicious-only artifact investigation scope.
raw_security = pd.read_csv("../data/endpoint_security.csv")

# Identify every endpoint present in the complete raw telemetry.
all_endpoints = sorted(
    raw_security["endpoint_id"]
    .dropna()
    .unique()
)

# Identify endpoints with at least one explicit FileType: Malicious record.
malicious_records = raw_security[
    raw_security["message"]
    .fillna("")
    .str.contains(
        "FileType: Malicious",
        case=False,
        na=False,
    )
]

malicious_endpoints = sorted(
    malicious_records["endpoint_id"]
    .dropna()
    .unique()
)

# Identify endpoints present in the raw data but without an explicit
# malicious security detection.
non_malicious_endpoints = sorted(
    set(all_endpoints) - set(malicious_endpoints)
)

print("=" * 60)
print("RAW DATASET ENDPOINT VERIFICATION")
print("=" * 60)

print(f"Total endpoints in raw dataset: {len(all_endpoints)}")
print(
    f"Endpoints with explicit malicious detections: "
    f"{len(malicious_endpoints)}"
)
print(
    f"Endpoints without explicit malicious detections: "
    f"{len(non_malicious_endpoints)}"
)

print("\nAll endpoints:")
print(all_endpoints)

print("\nEndpoints with explicit malicious detections:")
print(malicious_endpoints)

print("\nEndpoints without explicit malicious detections:")
print(non_malicious_endpoints)
```
